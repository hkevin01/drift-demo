# Drift Exploration Notebook

Explore data drift, concept drift, and model drift signals from the pipeline output.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

LOG_PATH = os.path.join('..', 'outputs', 'drift_log.csv')
df = pd.read_csv(LOG_PATH)
df.head()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 4))
ax[0].plot(df['batch'], df['accuracy'])
ax[0].set_title('Accuracy Over Time')
ax[1].plot(df['batch'], df['psi'], label='PSI')
ax[1].plot(df['batch'], df['kl_div'], label='KL')
ax[1].plot(df['batch'], df['ks_pvalue'], label='KS p-value')
ax[1].legend()
ax[1].set_title('Drift Signals')
plt.tight_layout()
```
